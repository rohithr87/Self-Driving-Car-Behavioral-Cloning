```python
# -*- coding: utf-8 -*-
"""somewhat_working.ipynb - MINIMALLY TUNED VERSION

Only critical minimal changes for better convergence
"""

from google.colab import drive
drive.mount('/content/drive')

import numpy as np
if not hasattr(np, 'sctypes'):
    np.sctypes = {
        'int': [np.int8, np.int16, np.int32, np.int64],
        'uint': [np.uint8, np.uint16, np.uint32, np.uint64],
        'float': [np.float16, np.float32, np.float64],
        'complex': [np.complex64, np.complex128],
        'others': [bool, object, bytes, str, np.void]
    }
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import keras
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten, ELU, Lambda
from keras.layers import Conv2D, BatchNormalization
from keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from keras.preprocessing import image
from keras.optimizers import Adam
import cv2
import pandas as pd
import os
import ntpath
from sklearn.utils import shuffle
from sklearn.model_selection import train_test_split
from google.colab import files
import tensorflow as tf
import zipfile

zip_path = '/content/drive/MyDrive/my_data.zip'  # adjust path
extract_path = '/content/my_data'                # where to extract

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Files extracted to:", extract_path)

# ===================== CONFIGURATION (MINIMAL CHANGES) =====================
BATCH_SIZE = 32  # Reduced from 64 for better gradient updates
EPOCHS = 30      # Increased from 20 for more training time
LEARNING_RATE = 1e-4

# ===================== DATA LOADING (UNCHANGED) =====================
datadir = '/content/my_data/my_data' # Corrected path based on file listing
columns = ['center', 'left', 'right', 'steering', 'throttle', 'reverse', 'speed']
# Skip the header row
data = pd.read_csv(os.path.join(datadir, 'driving_log.csv'), names=columns, skiprows=[0])

def path_leaf(path):
    head, tail = ntpath.split(path)
    return tail

data['center'] = data['center'].apply(path_leaf)
data['left'] = data['left'].apply(path_leaf)
data['right'] = data['right'].apply(path_leaf)

original_data = data.copy()

plt.figure(figsize=(15, 10))

# Before balancing
plt.subplot(2, 2, 1)
plt.hist(original_data['steering'], bins=100, color='blue', alpha=0.7)
plt.title('Steering Angle Distribution (Before Balancing)')
plt.xlabel('Steering Angle')
plt.ylabel('Count')
plt.grid(True, alpha=0.3)

# Before balancing (zoomed in)
plt.subplot(2, 2, 2)
plt.hist(original_data['steering'], bins=100, color='blue', alpha=0.7)
plt.title('Steering Angle Distribution - Zoomed (Before Balancing)')
plt.xlabel('Steering Angle')
plt.ylabel('Count')
plt.ylim(0, 200)  # Adjust this based on your data
plt.grid(True, alpha=0.3)

# ===================== DATA BALANCING (UNCHANGED) =====================
def balance_steering_data(data):
    """
    Balance steering angle data using a two-step process:
    1. Probabilistic reduction with 200 bins to bring bins to average
    2. Hard cap with 100 bins to limit to 100 samples per bin

    Parameters:
    data (DataFrame): Input data with 'steering' column (modified in place)
    """
    # STEP 1: Probabilistic balancing with 200 bins
    num_bins_prob = 200
    average_samples_per_bin = len(data) / num_bins_prob
    hist_prob, bins_prob = np.histogram(data['steering'], num_bins_prob)

    remove_list_prob = []
    for j in range(num_bins_prob):
        bin_indices = data.index[(data['steering'] >= bins_prob[j]) &
                                (data['steering'] <= bins_prob[j+1])].tolist()
        count = len(bin_indices)

        if count > average_samples_per_bin:
            keep_prob = average_samples_per_bin / count
            for idx in bin_indices:
                if np.random.rand() > keep_prob:
                    remove_list_prob.append(idx)

    data.drop(remove_list_prob, inplace=True)

    # STEP 2: Hard cap balancing with 100 bins
    num_bins_hard = 100
    hist_hard, bins_hard = np.histogram(data['steering'], num_bins_hard)

    remove_list_hard = []
    for j in range(num_bins_hard):
        bin_indices = data.index[(data['steering'] >= bins_hard[j]) &
                                (data['steering'] <= bins_hard[j+1])].tolist()

        if len(bin_indices) > 100:
            remove_indices = shuffle(bin_indices)[100:]
            remove_list_hard.extend(remove_indices)

    data.drop(remove_list_hard, inplace=True)

    return data

data = balance_steering_data(data)

plt.subplot(2, 2, 3)
plt.hist(data['steering'], bins=100, color='green', alpha=0.7)
plt.title('Steering Angle Distribution (After Balancing)')
plt.xlabel('Steering Angle')
plt.ylabel('Count')
plt.ylim(0, 200)  # Adjust this based on your data
plt.grid(True, alpha=0.3)

# After balancing (zoomed in)
plt.subplot(2, 2, 4)
plt.hist(data['steering'], bins=100, color='green', alpha=0.7)
plt.title('Steering Angle Distribution - Zoomed (After Balancing)')
plt.xlabel('Steering Angle')
plt.ylabel('Count')
plt.ylim(0, 200)  # Adjust this based on your data
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
print(f"Final number of samples: {len(data)}")

# steering smoothed (UNCHANGED)
data['steering_smoothed'] = data['steering'].rolling(window=3, min_periods=1).mean()

# ===================== DATA PREPARATION (UNCHANGED) =====================
def load_img_steering(datadir, data):
    image_path = []
    steering = []
    for i in range(len(data)):
        indexed_data = data.iloc[i]
        center_img = indexed_data[0]
        steering_angle = float(indexed_data['steering_smoothed'])

        # Use center, left and right cameras with steering correction
        image_path.append(os.path.join(datadir, 'IMG', center_img.strip()))
        steering.append(steering_angle)

        # Left camera with correction
        image_path.append(os.path.join(datadir, 'IMG', indexed_data[1].strip()))
        steering.append(steering_angle+0.25)

        # Right camera with correction
        image_path.append(os.path.join(datadir, 'IMG', indexed_data[2].strip()))
        steering.append(steering_angle-0.25)

    return np.array(image_path), np.array(steering)

image_paths, steerings = load_img_steering(datadir, data)
X_train, X_val, y_train, y_val = train_test_split(image_paths, steerings, test_size=0.2, random_state=5)

# ===================== AUGMENTATIONS (UNCHANGED) =====================
def add_random_shadow(image):
    """Add random shadow augmentation"""
    h, w = image.shape[0], image.shape[1]
    # Random shadow line coordinates
    x1, y1 = w * np.random.rand(), 0
    x2, y2 = w * np.random.rand(), h
    xm, ym = np.mgrid[0:h, 0:w]

    # Create mask based on line equation
    mask = np.zeros_like(image[:, :, 0])
    mask[(ym - y1) * (x2 - x1) - (y2 - y1) * (xm - x1) > 0] = 1

    # Apply shadow with random intensity
    cond = mask == 1
    s_variation = np.random.uniform(low=0.2, high=0.5)
    hls = cv2.cvtColor(image, cv2.COLOR_RGB2HLS)
    hls[:, :, 1][cond] = hls[:, :, 1][cond] * s_variation
    image = cv2.cvtColor(hls, cv2.COLOR_HLS2RGB)
    return image

def random_brightness(image):
    """Random brightness adjustment"""
    hsv = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)
    ratio = 1.0 + 0.4 * (np.random.rand() - 0.5)
    hsv[:, :, 2] = hsv[:, :, 2] * ratio
    return cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

def random_shift(image, steering):
    """Random horizontal and vertical shift for (128, 256) input images"""
    h, w = image.shape[:2]

    # Random shift amounts (in pixels)
    tx = np.random.randint(-40, 40)  # Horizontal shift: -40 to +40 pixels
    ty = np.random.randint(-10, 10)   # Vertical shift: -10 to +10 pixels

    # Create translation matrix
    M = np.float32([[1, 0, tx], [0, 1, ty]])

    # Apply shift to image
    shifted_image = cv2.warpAffine(image, M, (w, h))

    # Adjust steering angle based on horizontal shift
    # 0.004 is a common factor used in self-driving car projects
    steering += tx * 0.004

    # Ensure steering stays within valid range
    steering = np.clip(steering, -1.0, 1.0)

    return shifted_image, steering

def random_flip(image, steering):
    """Random horizontal flip"""
    if np.random.rand() < 0.5:
        image = cv2.flip(image, 1)
        steering = -steering
    return image, steering

def random_augment(image, steering):
    """Apply all augmentations randomly"""
    image = mpimg.imread(image) if isinstance(image, str) else image

     # Apply shift with probability (REDUCED from 0.8 to 0.5)
    if np.random.rand() < 0.5:
        image, steering = random_shift(image, steering)

    # Apply flip with probability (REDUCED from 0.6 to 0.5)
    if np.random.rand() < 0.5:
        image, steering = random_flip(image, steering)

    # Apply other augmentations with probability (REDUCED from 0.5 to 0.3)
    if np.random.rand() < 0.3:
        image = add_random_shadow(image)
    if np.random.rand() < 0.3:
        image = random_brightness(image)

    return image, steering

def nvidia_preprocess(image):
    """
    Preprocessing specifically for NVIDIA PilotNet model
    Based on the original paper implementation
    """
    # Crop to focus on road area (remove sky and car hood)
    # These values are approximate and should be adjusted for your camera setup
    image = image[60:135, :, :]

    # Resize to NVIDIA's standard input size
    image = cv2.resize(image, (200, 66))

    # Convert to YUV color space (as used in NVIDIA's implementation)
    image = cv2.cvtColor(image, cv2.COLOR_RGB2YUV)

    return image

# List the contents of the extracted directory
!ls /content/my_data
# If there are subdirectories, list their contents as well
!ls /content/my_data/*

# ===================== BATCH GENERATOR (UNCHANGED) =====================
def batch_generator(image_paths, steering_ang, batch_size, is_training):
    while True:
        batch_img = []
        batch_steering = []

        for i in range(batch_size):
            random_index = np.random.randint(0, len(image_paths))

            if is_training:
                img, steering = random_augment(image_paths[random_index], steering_ang[random_index])
            else:
                img = mpimg.imread(image_paths[random_index])
                steering = steering_ang[random_index]

            img = nvidia_preprocess(img)
            batch_img.append(img)
            batch_steering.append(steering)

        yield np.array(batch_img), np.array(batch_steering)

def visualize_batch_generator(generator, steps=1, figsize=(20, 15)):
    """
    Visualize multiple batches from the generator to inspect augmentations and data quality
    """
    for batch_num in range(steps):
        # Get a batch from the generator
        batch_images, batch_steerings = next(generator)

        # Calculate grid size
        batch_size = len(batch_images)
        cols = 8  # Number of columns in the grid
        rows = int(np.ceil(batch_size / cols))

        # Create figure
        plt.figure(figsize=figsize)
        plt.suptitle(f'Batch #{batch_num+1} - Size: {batch_size} - Steering Range: [{np.min(batch_steerings):.3f}, {np.max(batch_steerings):.3f}]',
                     fontsize=16, y=0.98)

        for i in range(batch_size):
            plt.subplot(rows, cols, i+1)

            # Display image (assuming RGB format)
            plt.imshow(batch_images[i])
            plt.title(f'Steering: {batch_steerings[i]:.3f}', fontsize=10)
            plt.axis('off')

            # Add a colored border based on steering angle
            steering = batch_steerings[i]
            if steering > 0.1:
                border_color = 'red'  # Right turn
            elif steering < -0.1:
                border_color = 'blue'  # Left turn
            else:
                border_color = 'green'  # Straight

            # Add border
            for spine in plt.gca().spines.values():
                spine.set_color(border_color)
                spine.set_linewidth(2)

        plt.tight_layout()
        plt.show()

        # Print batch statistics
        print(f"Batch #{batch_num+1} Statistics:")
        print(f"  Steering Mean: {np.mean(batch_steerings):.4f}")
        print(f"  Steering Std: {np.std(batch_steerings):.4f}")
        print(f"  Steering Min: {np.min(batch_steerings):.4f}")
        print(f"  Steering Max: {np.max(batch_steerings):.4f}")
        print(f"  Image Shape: {batch_images[0].shape}")
        print(f"  Image Range: [{np.min(batch_images[0])}, {np.max(batch_images[0])}]")
        print("-" * 50)

# Create a generator instance for visualization
vis_generator = batch_generator(X_train, y_train, BATCH_SIZE, True)

# Visualize 2 batches
visualize_batch_generator(vis_generator, steps=2)

# ===================== ORIGINAL NVIDIA MODEL (UNCHANGED) =====================
def nvidia_pilotnet_model(input_shape=(66, 200, 3)):
    """
    NVIDIA PilotNet model architecture as described in:
    "End to End Learning for Self-Driving Cars" by Bojarski et al.

    Parameters:
    input_shape (tuple): Expected input shape (height, width, channels)
                         Default is (66, 200, 3) as used in the original paper

    Returns:
    model: Compiled Keras model
    """
    model = Sequential()

    # Normalization layer - Standard practice in NVIDIA's implementation
    model.add(Lambda(lambda x: x/127.5 - 1.0, input_shape=input_shape))

    # Convolutional layers as described in the paper
    # First three layers use 5x5 kernels with 2x2 stride
    model.add(Conv2D(24, (5, 5), strides=(2, 2), activation='elu'))
    model.add(Conv2D(36, (5, 5), strides=(2, 2), activation='elu'))
    model.add(Conv2D(48, (5, 5), strides=(2, 2), activation='elu'))

    # Last two convolutional layers use 3x3 kernels with 1x1 stride
    model.add(Conv2D(64, (3, 3), activation='elu'))
    model.add(Conv2D(64, (3, 3), activation='elu'))

    # Flatten before fully connected layers
    model.add(Flatten())

    # Fully connected layers
    model.add(Dense(100, activation='elu'))
    model.add(Dense(50, activation='elu'))
    model.add(Dense(10, activation='elu'))

    # Output layer - single neuron for steering angle prediction
    model.add(Dense(1))

    # Compile with appropriate loss function and optimizer
    model.compile(
        optimizer=Adam(learning_rate=1e-4),
        loss='mse',
        metrics=['mae']
    )

    return model

# Create the model
model = nvidia_pilotnet_model(input_shape=(66, 200, 3))
print(model.summary())

# Calculate proper steps per epoch
train_steps = len(X_train) // BATCH_SIZE
val_steps = len(X_val) // BATCH_SIZE

# Enhanced callbacks (MINIMAL CHANGES - only patience increased)
callbacks = [
    EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True),  # Increased patience
    ModelCheckpoint('nvidia.h5', save_best_only=True),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-7)  # Increased patience
]

# Train the model
history = model.fit(
    batch_generator(X_train, y_train, BATCH_SIZE, True),
    steps_per_epoch=train_steps,
    epochs=EPOCHS,
    validation_data=batch_generator(X_val, y_val, BATCH_SIZE, False),
    validation_steps=val_steps,
    callbacks=callbacks,
    verbose=1
)

# ===================== VISUALIZATION & SAVING (UNCHANGED) =====================
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.legend()
plt.title('Loss Evolution')
plt.xlabel('Epoch')

model.save('nvidia.h5')
files.download('nvidia.h5')
```
